## 🔌 API Endpoints

| Method | Endpoint | Description |
|--------|----------|-------------|
| GET | `/api/home/` | Get menu, categories, offers, banners |
| GET | `/api/trending/` | Get trending items |
| GET | `/api/menu/filter/?category_id=1` | Filter menu by category |
| GET | `/api/tables/` | Get all tables with QR codes |
| POST | `/api/orders/` | Create order & Razorpay payment |
| POST | `/api/verify-payment/` | Verify payment signature |
| GET | `/api/invoice/<id>/` | Get order invoice |

## 📱 Complete User Flow

```
1. Customer scans QR on table
   ↓
2. Browser opens http://localhost:3000/table/1
   ↓
3. Table number (1) stored in localStorage
   ↓
4. Redirected to homepage
   ↓
5. View menu items & categories
   ↓
6. Add items to cart (only if table set)
   ↓
7. View cart with table badge
   ↓
8. Checkout
   ↓
9. Razorpay payment window opens
   ↓
10. Enter test card: 4111 1111 1111 1111
    ↓
11. Payment verified
    ↓
12. Order saved with table number
    ↓
13. Invoice generated
    ↓
14. Order appears in admin panel
```
